```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("painter")
```

```python
"""POS
noun - n 
verb- v 
adjective - a
adverb -r"""
lemmatizer.lemmatize("painting",pos='v')
```
