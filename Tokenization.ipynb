```python
corpus="""Hello, Welcome to krish naik Tutorials.
Pleasee do watch the entire course! to become an expert in NLP.
"""
```

```python
corpus
```

```python
print(corpus)
```

### Tokenization
### Paragraph to sentence

```python

from nltk.tokenize import sent_tokenize
import nltk
nltk.download('punkt')
```

```python
sent_tokenize(corpus)
```

```python
document = sent_tokenize(corpus)
```

```python
document
```

```python
type(document)
```

```python
for sentence in document:
    print(sentence)
```

### Pargarph inton wprds 
### sentences into words

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python

for sentence in document:
    print(word_tokenize(sentence))
```

```python
# Here even punctuation can also get seperated 
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
### Except last full stop , all other full stop will be considered nd taken in group with the last word it come along with 
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
